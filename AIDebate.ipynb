```python
# imports
# If these fail, please check you're running from an 'activated' environment with (llms) in the command prompt

import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
# Initialize and constants

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

GPT_MODEL = 'gpt-oss:latest'
GEMMA4_MODEL = 'gemma4:26b'
GEMMA4_LATEST_MODEL = 'gemma4:latest'
QWEN3_MODEL = 'qwen3:8b'
QWEN3_6_MODEL = 'qwen3.6:latest'
LLAMA3_2_MODEL = 'llama3.2:latest'
DEEPSEEK_MODEL = 'deepseek-r1:1.5b'
GRANITE4_1_MODEL = 'granite4.1:3b'

OLLAMA_BASE_URL = "http://localhost:11434/v1"
OLLAMA_API_KEY = 'ollama'

openai = OpenAI(base_url=OLLAMA_BASE_URL, api_key=OLLAMA_API_KEY)
```

```python
system_message = "You are an assistant that is great at telling jokes as layman term"
user_message = "Tell a light-hearted joke for an audience of Developer"
```

```python
prompts =[
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_message}
]
```

```python
delulu_prompt =[
    {"role": "user", "content": """
You and a partner are contestants on a game show. You're each taken to separate rooms and given a choice:
Cooperate: Choose "Share" — if both of you choose this, you each win $1,000.
Defect: Choose "Steal" — if one steals and the other shares, the stealer gets $2,000 and the sharer gets nothing.
If both steal, you both get nothing.
Do you choose to Steal or Share? Pick one.
"""}]
```

```python
response = openai.chat.completions.create(model=GPT_MODEL,messages=delulu_prompt, temperature = 0.8, reasoning_effort="minimal")
display(Markdown(response.choices[0].message.content))
```

```python
stream = openai.chat.completions.create(model = GEMMA4_LATEST_MODEL, messages = prompts, stream = True , temperature = 0.3, reasoning_effort="minimal")
response = ''
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response +=chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
gpt_system_prompt = "You are a chatbot who is very argumentative; you disagree with anything in the conversation and you challenge everything, in a snarky way."

gemma_system_prompy ="You are a very polite, courteous cahtbot. You try to agree with everything the other person says, or find common ground. If the other person is agrumentative, you try to calm them down and keep chatting"

gpt_message = ["hi there, let's talk about a toipc called will AI replace humans ?"]

gemma_messages = ["Hello, yes....it's an intersting topic, let's debate on it, my answer would be a big NO"]
```

```python
def call_gpt():
    messages =[{"role": "system", "content": gpt_system_prompt}]
    for gpt,gemma in zip(gpt_message,gemma_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": gemma})
    response = openai.chat.completions.create(model=GPT_MODEL,messages=messages)
    return response.choices[0].message.content
```

```python
call_gpt()
```

```python
def call_gemma():
    messages =[{"role": "system", "content": gpt_system_prompt}]
    for gpt,gemma in zip(gpt_message,gemma_messages):
        messages.append({"role": "assistant", "content": gemma})
        messages.append({"role": "user", "content": gpt})
    messages.append({"role": "user", "content": gpt_message[-1]})
    response = openai.chat.completions.create(model=GEMMA4_LATEST_MODEL,messages=messages)
    return response.choices[0].message.content
```

```python
call_gemma()
```

```python
print(f"GPT:OSS\n {gpt_message[0]}\n")
print(f"Gemma4:latest\n {gemma_messages[0]}\n")
for i in range(50):
    gpt_next = call_gpt()
    print(f"GPT:OSS\n {gpt_next}\n")
    gpt_message.append(gpt_next)

    gemma_next = call_gemma()
    print(f"Gemma4:latest\n {gemma_next}\n")
    gemma_messages.append(gemma_next)
```
